```python
import csv

rows = []
with open('../DADES_DEMOGRAFIQUES_VARIES - DADES_DEMOGRAFIQUES_VARIES.csv') as csvfile:
    readCSV = csv.reader(csvfile, delimiter=',')
    for row in readCSV:
        rows.append(row)
```

```python
fields = rows[0]
sm_nbhs = [r for r in rows if r[1] == '10']
[r[0] for r in sm_nbhs]
```

```python
nbhs_2_ids = {
    'el Parc i la Llacuna del Poblenou' : 66,
 "el Camp de l'Arpa del Clot" : 64,
 'la Vila Olímpica del Poblenou': 67,
 'el Clot' : 65,
 'el Poblenou' : 68,
 'Sant Martí de Provençals' : 72,
 'la Verneda i la Pau' : 73,
 'el Besòs i el Maresme' : 70,
}
```

```python
nbh_dict = {}
for nbh in sm_nbhs: 
    id = nbhs_2_ids[nbh[0]]
    nbh_dict[id] = {f : v for f,v in zip(fields, nbh)}


```

```python
import copy
nbh_dict[69] = copy.deepcopy(nbh_dict[70])
nbh_dict[69]["NOMBRE"] =  "Diagonal Mar i el Front Marítim"
nbh_dict[71] = copy.deepcopy(nbh_dict[72])
nbh_dict[71]["NOMBRE"] =  "Provençals de Poblenou"
```

```python
import json
with open("../barris.json", "w") as json_file:
    json.dump(nbh_dict, json_file)
```
